```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [5, 1, 7, 8, 2]
```

```python
plt.plot(x, y, color = "red", marker = "o")
plt.title("line plot to understand the syntax")
plt.xlabel("x points")
plt.ylabel("y points")
plt.show()
```

```python
plt.plot(x, y, color = "red", marker = "o", linestyle = '--')
plt.title("line plot to understand the syntax")
plt.xlabel("x points")
plt.ylabel("y points")
plt.show()
```

```python
plt.figure()
plt.subplot(2, 2, 2)
plt.plot(x, y)


plt.subplot(2, 2, 1)
plt.plot(x, y)

plt.subplot(2, 2, 3)
plt.plot(x, y)

plt.show()
```

```python
import numpy as np

plt.figure()
plt.subplot(2, 2, 1)
plt.plot(x, np.sin(x))


plt.subplot(2, 2, 2)
plt.plot(x, np.cos(x))

plt.subplot(2, 2, 3)
plt.plot(x, np.tan(x))

plt.subplot(2, 2, 4)
plt.plot(x, y)

plt.show()
```

```python
x = [8, 4, 9, 3, 6, 3]
y1 = [6, 7, 8, 9, 1, 1]
y2 = [8, 4, 9, 3, 6, 3]

plt.plot(x, y1, color = 'red', marker = 'o', markersize = '5', linestyle = '-', label = 'Y1 Line', linewidth = 2)
plt.plot(x, y2, color = 'blue', marker = 'o', markersize = '4', linestyle = '-', label = 'Y2 Line', linewidth = 2)

plt.legend()
plt.title('Line Chart')
plt.xlabel('X point')
plt.ylabel('Y point')
plt.xlim(1 , 10)
plt.ylim(0, 12)
#plt.grid()
plt.show()
```
