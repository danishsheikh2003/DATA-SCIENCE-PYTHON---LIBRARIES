```python
import matplotlib.pyplot as plt
```

```python
x = [1, 2, 3, 4, 5]
y = [5, 1, 7, 8, 2]
```

```python
plt.plot(x, y, color = "red", marker = "o")
plt.title("line plot to understand the syntax")
plt.xlabel("x points")
plt.ylabel("y points")
plt.show()
```

```python
plt.plot(x, y, color = "red", marker = "o", linestyle = '--')
plt.title("line plot to understand the syntax")
plt.xlabel("x points")
plt.ylabel("y points")
plt.show()
```

```python
plt.figure()
plt.subplot(2, 2, 2)
plt.plot(x, y)


plt.subplot(2, 2, 1)
plt.plot(x, y)

plt.subplot(2, 2, 3)
plt.plot(x, y)

plt.show()
```

```python
import numpy as np

plt.figure()
plt.subplot(2, 2, 1)
plt.plot(x, np.sin(x))


plt.subplot(2, 2, 2)
plt.plot(x, np.cos(x))

plt.subplot(2, 2, 3)
plt.plot(x, np.tan(x))

plt.subplot(2, 2, 4)
plt.plot(x, y)

plt.show()
```

```python
data1 = np.random.normal(0, 1, 5000)
data2 = np.random.normal(3, 2, 5000)

plt.hist([data1, data2], bins = 20, color = ['r', 'b'], label = ['dist1', 'dist2'])
plt.xlabel('Values')
plt.ylabel('frequency')
plt.legend() #plt.legend()
plt.title("stacked histogram")
plt.show()
```
