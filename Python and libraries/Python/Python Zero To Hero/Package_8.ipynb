```python
import csv 

data = [
    ["Name", "Age", "Marks"],
    ["Danish", "22", "449"],
    ["sohel", "23", "478"],
    ["Mubashishir", "23", "498"]
]

with open("Student.csv", "w") as d:
    write = csv.writer(d)

    for i in data:
        write.writerow(i)  

with open("Student.csv", "r") as d:
    read = csv.reader(d)

    for i in read:
        print(i)
```

```python
import json

data = {
    "Name":"Danish",
    "Age" : 22,
    "Marks":449,
}

data1 = {
    "Name":"Sohel",
    "Age" : 23,
    "Marks":419,
}

data2 = {
    "Name":"Mubashishir",
    "Age" : 23,
    "Marks":498
}

with open("Student.json", "w") as d:
    json.dump((data, data1, data2), d)

with open("Student.json", "r") as d:
    line = json.load(d)
print(line)      
```

```python
import logging

# Logging configuration
logging.basicConfig(
    filename="Debug.log",
    level=logging.DEBUG,
    format="%(levelname)s - %(asctime)s - %(message)s",
    force=True
)

try:
    # Taking input
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))

    logging.debug("Input numbers received successfully")

    # Addition
    result = num1 + num2

    logging.info("Addition performed successfully")

    # Print result
    print("Addition:", result)

except ValueError:
    logging.error("Invalid input! Please enter numeric values.")
    print("Error: Please enter valid numbers.")

finally:
    logging.shutdown()
```
