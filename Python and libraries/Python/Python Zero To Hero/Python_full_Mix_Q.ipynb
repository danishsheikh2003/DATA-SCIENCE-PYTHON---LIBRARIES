```python
numbers = [10, -5, 0, 25, -15, 0, 30, -2]

def analyze_numbers(numbers):
    Zero = 0
    Positive = 0
    Negative = 0

    for i in numbers:
        if i > 0:
            Positive += 1
            
        elif i < 0:
            Negative += 1
            
        else:
            Zero += 1

    return {
        "positive": Positive,
        "negative": Negative,
        "zero": Zero
    }    
            
result = analyze_numbers(numbers)      
print(result)
```

```python
def reverse_string(text):
    result = ""
    for i in text:
        result = i + result
    return result

print(reverse_string("Python"))
```

```python
New = [10, 45, 23, 89, 12]



def find_max(New):
    
    idx = New[0]
    
    for i in New:
        if i > idx:
            idx = i 
    return idx   

find_max(New)    
```

```python
numbers = [10, 15, 20, 25, 30, 35]

def even_numbers(numbers):
    data =[]
    for i in numbers:
        if i%2 == 0:
            data.append(i)
    return data
    
even_numbers(numbers)        
```

```python
def count_vowels(text):
    count = 0    
    for i in text:
        if i in 'aeiouAEIOU':
            count += 1
    return ("Total Voval =",count)  

count_vowels("DataScience")    
```

```python
students = {"Amit": 80, "Raj": 40, "Sam": 65, "Ali": 30}

def Greater(data):
    name = []
    for Name,Marks in data.items():
        if Marks > 50:
            name.append(Name)
    return name
    
Greater(students)
```
