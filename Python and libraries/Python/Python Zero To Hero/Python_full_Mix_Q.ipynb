```python
numbers = [10, -5, 0, 25, -15, 0, 30, -2]

def analyze_numbers(numbers):
    Zero = 0
    Positive = 0
    Negative = 0

    for i in numbers:
        if i > 0:
            Positive += 1
            
        elif i < 0:
            Negative += 1
            
        else:
            Zero += 1

    return {
        "positive": Positive,
        "negative": Negative,
        "zero": Zero
    }    
            
result = analyze_numbers(numbers)      
print(result)
```

```python
from functools import reduce


# DECORATOR
def order_status(func):
    def wrapper(self):
        print("\nProcessing Order...")
        result = func(self)
        print("Order Successful!")
        return result
    return wrapper


# CUSTOMER CLASS
class Customer:

    def __init__(self, name, customer_id, city):
        self.name = name
        self.customer_id = customer_id
        self.city = city

    def show_details(self):
        return {
            "Name": self.name,
            "Customer ID": self.customer_id,
            "City": self.city
        }


# PRODUCT CLASS
class Product:

    def __init__(self, product_name, price, quantity):
        if price <= 0 or quantity <= 0:
            raise ValueError("Invalid price or quantity")

        self.product_name = product_name
        self.price = price
        self.quantity = quantity

    def show_product(self):
        return f"{self.product_name} - ₹{self.price}"

    def update_quantity(self, quantity):
        if quantity < 0:
            raise ValueError("Invalid quantity")

        self.quantity = quantity


# ORDER CLASS
class Order:

    def __init__(self, order_id, customer, discount=0):
        self.order_id = order_id
        self.customer = customer
        self.products = []
        self.discount = discount

    # Add product using *args
    def add_product(self, *products):
        for product in products:
            self.products.append(product)

    # Remove product
    def remove_product(self, product_name):
        self.products = [
            p for p in self.products
            if p.product_name != product_name
        ]

    # Calculate total using reduce
    def calculate_total(self):

        if not self.products:
            raise ValueError("Order is empty")

        prices = [
            p.price * p.quantity
            for p in self.products
        ]

        return reduce(lambda a, b: a + b, prices)

    # Apply discount
    def apply_discount(self):

        total = self.calculate_total()

        if self.discount < 0 or self.discount > total:
            raise ValueError("Invalid discount")

        return total - self.discount

    # Filter premium products
    def premium_products(self):

        return list(
            filter(
                lambda p: p.price > 500,
                self.products
            )
        )

    # Generator
    def product_generator(self):

        for product in self.products:
            yield product

    # Display order
    @order_status
    def show_order(self):

        print("\n========== ORDER SUMMARY ==========")

        print("\nCustomer Details:")
        print(self.customer.show_details())

        print("\nOrder ID:", self.order_id)

        print("\nProducts:")

        for product in self.product_generator():
            print(product.show_product())

        total = self.calculate_total()
        final_bill = self.apply_discount()

        print("\nPremium Products:")

        for product in self.premium_products():
            print(product.product_name)

        print("\nTotal Bill:", total)
        print("Discount:", self.discount)
        print("Final Bill:", final_bill)


# CREATE CUSTOMER
customer = Customer("Danish", 101, "Nagpur")


# CREATE PRODUCTS
p1 = Product("Mobile", 12000, 1)
p2 = Product("Mouse", 800, 1)
p3 = Product("Keyboard", 1200, 1)


# CREATE ORDER
order = Order(5001, customer, discount=1000)


# ADD PRODUCTS
order.add_product(p1, p2, p3)


# DISPLAY ORDER
order.show_order()
```

```python
class Bank:

    def __init__(self, name, balance):
        self.name = name
        self.__balance = balance

        self.starting_balance = balance
        self.result1 = 0
        self.result = 0

        self.history = []

    def check_balance(self):
        return self.__balance

    def deposit(self, amount):
        if amount < 0:
            raise ValueError("Amount Cant be Negative")
        self.__balance += amount
        self.result1 += amount

        self.history.append(f"Deposited: ₹{amount}")

    def withdraw(self,amount1):
        if amount1 < 0: 
            raise ValueError("Amount Cant be Negative")

        if amount1 > self.__balance:
            raise ValueError("Insufficient balance")
            
        if amount1 <= self.__balance:     # 20,000 < 30,000
            self.__balance -= amount1
            self.result += amount1

            self.history.append(f"Withdrawn: ₹{amount1}")
            
    def show_details(self):
        print("====== Bank Transaction ======\n")
        print(f"Costomer : {self.name}")
        print(f"Starting balance : {self.starting_balance}\n")

        print(f"Deposited : {self.result1}")
        print(f"Withdraw : {self.result}\n")

        print(f"Final Balance: {self.__balance}")

    def transaction_history(self):
        print("\n===== Transaction History =====")

        for transaction in self.history:
            print(transaction)
         
         

bank = Bank('Danish', 45000)

bank.deposit(5000)

bank.check_balance()

bank.withdraw(25000)

bank.check_balance()

bank.show_details()

bank.transaction_history()
```

```python

from functools import reduce


class Student:

    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = list(marks)

        self.total = 0
        self.percentage = 0
        self.grade = ""

    def calculate_total(self):
        self.total = reduce(lambda a, b: a + b, self.marks)
        return self.total

    def calculate_percentage(self):
        self.percentage = self.total / (len(self.marks) * 100) * 100
        return self.percentage

    def calculate_grade(self):

        if self.percentage >= 90:
            self.grade = "Grade A"
        elif self.percentage >= 75:
            self.grade = "Grade B"
        elif self.percentage >= 60:
            self.grade = "Grade C"
        elif self.percentage >= 40:
            self.grade = "Grade D"
        else:
            self.grade = "Fail"

        return self.grade

    def show_result(self):
        print("\n===== Student Result =====")
        print(f"Name: {self.name}")
        print(f"Roll No: {self.roll_no}")
        print(f"Total Marks: {self.total}")
        print(f"Percentage: {self.percentage:.2f}%")
        print(f"Grade: {self.grade}")


# Creating student objects
student = Student("Mubashishir", 100, [99, 79, 98])
student1 = Student("Danish", 101, [80, 75, 90])
student2 = Student("Rahul", 102, [60, 65, 70])
student3 = Student("Aman", 103, [30, 35, 40])

# Store all students in a list
students = [student, student1, student2, student3]


# Calculate and display all results
for stud in students:
    stud.calculate_total()
    stud.calculate_percentage()
    stud.calculate_grade()
    stud.show_result()


# Find class topper
topper = max(students, key=lambda s: s.total)

print("\n===== Class Topper =====")
print(f"Name: {topper.name}")
print(f"Total Marks: {topper.total}")


# Find failed students
failed_students = [s for s in students if s.percentage < 40]

print("\n===== Failed Students =====")

if failed_students:
    for s in failed_students:
        print(f"Name: {s.name}")
        print(f"Percentage: {s.percentage:.2f}%")
else:
    print("No Failed Students")


# Search student by roll number
search_roll = int(input("\nEnter Roll Number [100, 101, 102, 103]: "))

found = False

for s in students:
    if s.roll_no == search_roll:
        s.show_result()
        found = True
        break

if not found:
    print("Student Not Found")
```
