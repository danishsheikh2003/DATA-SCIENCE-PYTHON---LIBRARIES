```python
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("Function starting...")
        result = func(*args, **kwargs)
        print("Function completed!")
        return result
    return wrapper

@my_decorator
def add(a, b):
    return a + b

@my_decorator
def multiply(a, b, c):
    return a * b * c

print(add(10, 20))
print(multiply(2, 3, 4))
```

```python
class BankAccount:

    def __init__(self, holder_name, balance):
        self.holder_name = holder_name
        self.__balance = balance

    @property
    def balance(self):
        return self.__balance

    @balance.setter
    def balance(self, amount):
        if amount < 0:
            raise ValueError("Balance can't be negetive")
        self.__balance = amount

    def deposit(self, amount):
        if amount <= 0:
            raise ValueError("Deposit can't be negetive")
        self.__balance = self.__balance + amount

    def withdraw(self, amount):
        if amount < 0:
            raise ValueError("Withdraw can't be negetive")

        if amount > self.__balance:
            raise ValueError("Insufficient balance")  
            
        self.__balance = self.__balance - amount 

    @balance.deleter
    def balance(self):
        del self.__balance

account = BankAccount("Danish", 100000)

#print(account.balance)  

account.deposit(250000)
print(account.balance)  

#account.withdraw(100000)
#print(account.balance)   
```

```python
class Product:

    def __init__(self, name, price):
        self.name = name
        self.__price = price

    @property
    def price(self):  
        return self.__price

    @price.setter
    def price(self, new_price): 
        if new_price < 0:
            raise ValueError("Balance can't be negetive")
        self.__price = new_price

    @price.deleter
    def price(self):
        del self.__price

    def apply_discount(self, percentage):
        if percentage <= 0 or percentage >= 100:
            raise ValueError("Discount must be between 0 and 100")
            
        result = self.price - (self.price / 100 * percentage)

        self.price = result

product = Product("Laptop", 50000)

print(product.price)       # 50000

product.price = 60000
print(product.price)       # 60000

product.apply_discount(10)
print(product.price)       # 54000            
```
