```python
import pandas as pd

df = pd.DataFrame({
    'Name': ['Amit', 'Rahul', 'Sahil', 'Neha', 'Priya'],
    'Age': [22, 25, 23, 26, 24],
    'Department': ['IT', 'HR', 'IT', 'Sales', 'IT'],
    'Salary': [25000, 30000, 28000, 35000, 27000]
})

df.loc[1, 'Salary'] = None

df['Salary'].isnull().sum()

df['Salary'] = df['Salary'].fillna(df['Salary'].mean())
```

```python
df[df['Department'].duplicated()]
```

```python
da1 = df.drop_duplicates(['Department'])
```

```python
da1
```

```python
import numpy as np
df.groupby(['Department'])['Salary'].agg(['mean', 'max', 'sum'])
```

```python
df.groupby('Department').agg(
    Average_Salary=('Salary', 'mean'),
    Maximum_Salary=('Salary', 'max'),
    Total_Employees=('Name', 'count')
)
```
